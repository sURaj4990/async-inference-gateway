```python
def sync_countdown(n: int) -> list[int]:
    result = []
    for i in range(n):
        time.sleep(1.0)
        result.append(i)
    return result

for i in sync_countdown(5):
    print(i)
```

```python
def sync_countdown(n: int):
    for i in range(n):
        time.sleep(1.0)
        yield i

for i in sync_countdown(5):
    print(i)
```

```python
async def async_countdown(n: int):
    for i in range(n):
        await asyncio.sleep(1.0)
        yield i

async for i in (async_countdown(5)):
    print(i)
```

```python
async def async_countdown(n: int):
    for i in range(n):
        await asyncio.sleep(1.0)
        yield i

async def main():
    async for i in async_countdown(5):
        print(i)

if __name__ == "__main__":
    asyncio.run(main())
```

```python
# The Bucket Problem
capacity = 3
drop_rate = 0.5
current_tokens = 0

t_last_accessed = 10
t_now = 10

delta_t = t_now - t_last_accessed

tokens = min(capacity, current_tokens + (delta_t * drop_rate))

if tokens >= 1:
    tokens -= 1
    t_last_accessed = t_now
    print(f"Available tokens: {tokens}")
else:
    print("Inssuficient tokens")
```

```python
# De-loading task to threads
import queue
import time

def producer_thread(q: queue.Queue, stop: object):
    for i in range(1, 6):
        time.sleep(0.5)
        q.put(f"item_{i}")
    q.put(stop)
```

```python
import threading


q = queue.Queue()
STOP = object()

worker = threading.Thread(
    target=producer_thread,
    args=(q, STOP)
)
worker.start()

while True:
    item = q.get()
    if item is STOP:
        break
    print(item, flush=True)
```
